# Imports

```python
from sc_workflows.preprocessing import normalize_counts
import scanpy as sc
```

# Load Data

```python
adata_filtered = sc.read_h5ad("data/pbmc3k_filtered")
adata_normalized = normalize_counts(
    adata_filtered,
    target_sum=1e4,
    log1p=True,
)

print(adata_normalized)
print(adata_normalized.layers.keys())
```

```python
import numpy as np

original = adata_filtered.X
preserved = adata_normalized.layers["counts"]

if hasattr(original, "toarray"):
    original = original.toarray()
    preserved = preserved.toarray()

assert np.array_equal(original, preserved)
```
